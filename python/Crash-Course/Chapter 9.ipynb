```python
#9.5 Oh shit now it's time to define the undefinable, a user.

class User():
    def __init__(self, first_name, last_name, age, specie):
        self.first_name = first_name
        self.last_name = last_name
        self.age = age
        self.specie = specie
        self.login = 0


    def describe_user(self):
        print(f""" The User is called {self.first_name} {self.last_name}
        This peculiar creature claims to be {self.specie} and {self.age} But I don't really buy it""")


    def greet_user(self):
        print(f"Hi there, dear {self.first_name}. Take your damn meds, you are talking to a computer and that's not normal")

    def login_attempt(self):
        #Added a new attempt at logging for the user
        self.login +=1
        print(f"This is attempt number{self.login}")
    def reset_login(self):
        #This resets the logins attempts you see
        self.login = 0
        print(f"We have reseted the logins attempts")         


Marcelo = User('Marcelo','Eduardo', 'Age','Human')
Nobby = User('Nobby', 'Nobs', '?', 'Nobby')

Marcelo.greet_user()
Marcelo.describe_user()
Nobby.greet_user()
Nobby.describe_user()
Nobby.login_attempt()
Nobby.login_attempt()
Nobby.login_attempt()
Nobby.login_attempt()
Nobby.reset_login()
             
```

```python
#9.7 Okay now we are gonna make a special subuser called admin you see

class Admin(User):
    def __init__(self, first_name, last_name, age, specie,privileges):
        super().__init__(first_name,last_name,age,specie)
        self.privileges = privileges


    def show_privies(self):
        print(f"The :{self.first_name} is and admin, he has the following priviligies")
        for privi in self.privileges:
            print(f"{privi}")


God = Admin(
'God',
'Big Man',
'Great Big Guy in The Sky',
'He is god duh',
['Do whatever he wants', 'Delete the world', 'Make another world', 'Rewrite the terminal', 'Delete users: from existance']

)

God.show_privies()
```

```python
#9.8 Okay man I think this god guy has far too many privileges. I  think his powers should be defined in another class alotogether.

class Priviliges():
    "A simple attempt at modelling priviliges of  users. Like God"
    def __init__(self, privileges, rank = 0):
       self.rank = rank
       self.privileges = privileges
    def show_privileges(self):
        print(f"""This admin is rank {self.rank} his priviliges are""")
        for privilege in self.privileges:
            print(f"{privilege}")



class Admin(User):
    def __init__(self, first_name, last_name, age, specie,privileges):
        super().__init__(first_name,last_name,age,specie)
        self.privileges = Priviliges(privileges)     


God = Admin(
    'God',
    'Big Man',
    'Great Big Guy in The Sky',
    'He is god duh',
    [
        'Do whatever he wants',
        'Delete the world',
        'Create another world',
        'Rewrite the terminal'
    ]
)

God.privileges.show_privileges()
```
