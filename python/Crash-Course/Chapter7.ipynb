```python
#7.8 Okay now it's all about delicious sandwhichs man, that's great 
sandwich_orders = ['Hamburger', 'Philly Cheese', 'Hot Dog']
finished_sandwichs = []
while sandwich_orders:
    new_sandwich = sandwich_orders.pop()
    print(f"Hey man here is you new sandwich, it's a delicious {new_sandwich}")
    finished_sandwichs.append(new_sandwich)

cn= 1
for sandwich in finished_sandwichs:
    print(f"Client n {cn} Oh man, that {sandwich} sure was delicious")   
    cn +=1 
```
