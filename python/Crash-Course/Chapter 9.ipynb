```python
#9.5 Oh shit now it's time to define the undefinable, a user.

class User():
    def __init__(self, first_name, last_name, age, specie):
        self.first_name = first_name
        self.last_name = last_name
        self.age = age
        self.specie = specie
        self.login = 0


    def describe_user(self):
        print(f""" The User is called {self.first_name} {self.last_name}
        This peculiar creature claims to be {self.specie} and {self.age} But I don't really buy it""")


    def greet_user(self):
        print(f"Hi there, dear {self.first_name}. Take your damn meds, you are talking to a computer and that's not normal")

    def login_attempt(self):
        #Added a new attempt at logging for the user
        self.login +=1
        print(f"This is attempt number{self.login}")
    def reset_login(self):
        #This resets the logins attempts you see
        self.login = 0
        print(f"We have reseted the logins attempts")         


Marcelo = User('Marcelo','Eduardo', 'Age','Human')
Nobby = User('Nobby', 'Nobs', '?', 'Nobby')

Marcelo.greet_user()
Marcelo.describe_user()
Nobby.greet_user()
Nobby.describe_user()
Nobby.login_attempt()
Nobby.login_attempt()
Nobby.login_attempt()
Nobby.login_attempt()
Nobby.reset_login()
             
```

```python
#9.6 Now are gonna model an ice cream stand, ain't that really fun. Let's start bycopying the stuf above we defined fro the reasturant, okay.

#9.4 now we are gonna soem more stuff to the restaurant class 

class Restaurant():
    "Let's define a really cool restaurant you see"
    def __init__(self,name,cuisine):
        self.name = name
        self.cuisine = cuisine
        self.n_served = 0

    def describe(self):
        print(f" We are the {self.name.title()} restaurant. We serve the best {self.cuisine} cuisine")
              
    def open(self):
        print(f"We are currently open")

    def update(self,customers_served):
        if customers_served >= self.n_served:
            self.n_served = customers_served
        else:
            print(f"You can't unserve a customer, you moron")
    def print_customers(self):
        print(f"We have served {self.n_served} customers so far")   


class icecream_stand(Restaurant):
    def __init__(self,name,cuisine,flavors):
        super().__init__(name,cuisine)
        self.flavors = flavors

    def show_flavors(self):
        print(f"This ice cream stands serves a plethora of delicious flavors, to freeze your soul such as:") 
        for flavor in flavors:
            print(f"{flavor}")   


JackFrost = icecream_stand('Jack Frost',
'Ice Cream Stand',
['Strawberry', 'Vanilla', 'Passion Fruit', 'Chocolate', 'Jacks special freeze mind'] )

JackFrost.show_flavors()
```
