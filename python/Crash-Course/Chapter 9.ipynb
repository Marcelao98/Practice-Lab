```python
#9.5 Oh shit now it's time to define the undefinable, a user.

class User():
    def __init__(self, first_name, last_name, age, specie):
        self.first_name = first_name
        self.last_name = last_name
        self.age = age
        self.specie = specie
        self.login = 0


    def describe_user(self):
        print(f""" The User is called {self.first_name} {self.last_name}
        This peculiar creature claims to be {self.specie} and {self.age} But I don't really buy it""")


    def greet_user(self):
        print(f"Hi there, dear {self.first_name}. Take your damn meds, you are talking to a computer and that's not normal")

    def login_attempt(self):
        #Added a new attempt at logging for the user
        self.login +=1
        print(f"This is attempt number{self.login}")
    def reset_login(self):
        #This resets the logins attempts you see
        self.login = 0
        print(f"We have reseted the logins attempts")         


Marcelo = User('Marcelo','Eduardo', 'Age','Human')
Nobby = User('Nobby', 'Nobs', '?', 'Nobby')

Marcelo.greet_user()
Marcelo.describe_user()
Nobby.greet_user()
Nobby.describe_user()
Nobby.login_attempt()
Nobby.login_attempt()
Nobby.login_attempt()
Nobby.login_attempt()
Nobby.reset_login()
             
```

```python
#9.9 Okay so now its time to think about cars once again huh
class Car():
    "A sim plee way to try and define  a car"


    def __init__(self,make,model,year):
        self.make =make
        self.model = model
        self.year = year
        self.odometer_reading = 0

    def get_descriptive_name(self):
        long_name = str(self.year)+''+self.make+self.model
        return long_name.title()

    def read_odometer(self):
        print("This car has"+ str(self.odometer_reading) + "miles on it")

    def update_odometer(self,mileage):
        if mileage >= self.odometer_reading:
            self.odometer_reading = mileage
        else:
            print("You can't roll back and odometer")

    def increment_odometer(self,miles):
        self.odometer_reading += miles

class Battery():
    "An attempt to model the battery of an electric car"
    def __init__(self,battery_size = 70):
        self.battery_size = battery_size

    def describe_battery(self):
        "Pritn a statement describing the battery size"
        print("This car has a"+ str(self.battery_size) + "-kWh battery")

    def get_range(self):
        """ Print a statement about the range  this battery provides"""
        if self.battery_size == 70:
            range = 240
        elif self.battery_size == 85:
            range =270

        message = "This car can go approximately" + str(range)
        message += "miles on a full charge"

        print(message) 

    def upgrade_Battery(self):
        "Upgrade a statement about the range of the battery"
        if self.battery_size != 85:
            self.battery_size = 85

             

                

class ElectricCar(Car):
    """Represents aspects of cars specific to electric vehicles."""

    def __init__(self, make, model, year):
        """Initialize attributes of the parent class.
        Then initialize attributes specific to an electric car.
        """
        super().__init__(make, model, year)
        self.battery = Battery()




my_tesla = ElectricCar('tesla', 'model s', 2016)
print(my_tesla.get_descriptive_name())
my_tesla.battery.describe_battery()
my_tesla.battery.get_range()
my_tesla.battery.upgrade_Battery()
my_tesla.battery.get_range()


          
```
