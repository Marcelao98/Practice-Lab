```python
#8.13 Okay user profile 
def build_profile(first, last, **user_info):
    profile = {}
    profile['first_name'] = first
    profile['last_name'] = last
    for key, value in user_info.items():
        profile[key] = value 
    return profile 


user_profile = build_profile('Marcelo','Eduardo', location ='Igarapava', field = 'Electrical Engineering', hobbies = 'Tabletop RPG', age = '28', situation = 'unemployed, single, living with his mom. life is never been better though')

print(user_profile)


```

```python
#8.14 Alright the function is abotu cars and dicitonaries huh 

def make_car(manufacturer, model, **car_info):
    car = {}
    car['manufacturer_name'] = manufacturer
    car['model_name'] = model
    for key, value in car_info.items():
        car[key] = value
    return car




moms_car = make_car('Ford', 'Ka', Year = '2017', Color = 'White')

print(moms_car)

```
